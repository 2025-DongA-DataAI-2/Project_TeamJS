```python
# flask
from flask import render_template
from flask import request

# 시각화
import numpy as np
import matplotlib.pyplot as plt
```

```python
app = Flask(__name__)

@app.route("/")
def index():
    return render_template("cook.html")

@app.route("/main")
def main():
    return render_template("cook.html")

@app.route("/result")
def result():
    no = request.args.get('no', type=str)
    return render_template("page"+no+".html")

app.run(host="127.0.0.1", port=9000, debug=True, use_reloader=False)
```
